### Tensor datatypes

###### **Note:** Tensor datatypes is one of the 3 big errors you'll run into with PyTorch & deep learning:



1.   Tensors not right datatype
2.   Tensors not right shape
3.   Tensors not on right device



```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0,4.0,8.0],
                               dtype=None)
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensor = torch.tensor([3.0,4.0,8.0],
                               dtype=torch.float16)
float_16_tensor
```

```python
float_16_tensor.dtype
```

```python
float_32_tensor_test = torch.tensor([3.0,4.0,8.0],
                               dtype=None, #
                              device=None,
                              requires_grad=False)
float_32_tensor_test
```

```python
converter_float_32_to_16_tensor = float_32_tensor.type(torch.float16)
converter_float_32_to_16_tensor
```

```python
float_16_tensor*float_32_tensor
```

```python
int_32_tensor = torch.tensor([3,6,9],
                             dtype=torch.int32)
int_32_tensor
```

```python
int_32_tensor*float_32_tensor
```

### Getting information from tensors
######

1.   Tensors not right datatype - to do get datatype from a tensor can use `tensor.dtype`
2.   Tensors not right shape - to get shape from a tensor, can use `tensor.shape`
3.   Tensors not on right device - to get device from a tensor can use `tensor.device`
   



```python
# Create a Tensor
some_tensor = torch.rand(3,4)
some_tensor
```

```python
# Find out details about some tensor
print(some_tensor)
print(f"Dtatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```

```python
some_tensor.size(), some_tensor.shape
```

### Matrix multiplication

###### Two main ways of performing multiplication in  nuural networks and deep learning:

1. Element-wise multiplication
2. Matrix multiplication (dot product)

```python
# Elent wise multiplication
print(tensor, "*", tensor)
print(f"Eqals: {tensor*tensor}")
```

```python
# Matrix Multiplication

torch.matmul(tensor,tensor)
```

```python
# Matrix multiplication by hand
1*1 + 2*2 + 3*3
```
